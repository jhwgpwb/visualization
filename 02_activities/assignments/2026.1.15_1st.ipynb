```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(0)
x = np.arange(50)
y=np.random.randint(0,100,50)
```

```python
fig, ax =plt.subplots(figsize = (5,3))
ax.scatter (x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
```

```python
fig, ax =plt.subplots(figsize = (5,3))
ax.plot (x,y)
```

```python
fig, ax =plt.subplots(figsize = (5,3))
ax.hist (y,edgecolor='black')
ax.set_title("Histogram")
ax.set_ylabel("Growth")
ax.set_xlabel("...")
fig.tight_layout()
```

```python
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
font1 = {'family':'sans-serif','color':'navy','size':20}
font2 = {'family':'monospace','color':'green','size':14}


```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,
           y,
          marker='o',         
          color = "indigo")   
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,
        y,
        marker='o', 
        color = 'indigo', 
        linestyle = '--',   
        linewidth = 1)      
fig.show()

```
